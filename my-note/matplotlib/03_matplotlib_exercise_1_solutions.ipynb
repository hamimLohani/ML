### Problem 2: Basic Object-Oriented Line Plot

Create a 1D array `x` with 100 points from 0 to $2\pi$. Compute $y = \sin(x)$.
Using the Object-Oriented interface (`plt.subplots()`), plot $y$ vs $x$ with:
- Figure size `(8, 4)` and DPI 100
- Blue solid line with `linewidth=2`
- Title: `'Sine Function'`
- X-axis label: `'x (radians)'`
- Y-axis label: `'sin(x)'`
- Grid enabled with dashed lines and 0.5 alpha


```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(x, y, color='#1f77b4', linewidth=2, label=r'$\sin(x)$')
ax.set_title('Sine Function', fontsize=12, fontweight='bold')
ax.set_xlabel('x (radians)')
ax.set_ylabel('sin(x)')
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend()
plt.show()
```

### Problem 3: Multiple Lines with Custom Linestyles & Legend

On a single Axes, plot three curves over $x \in [0, 10]$:
1. $y_1 = \sin(x)$: Solid line (`'-'`), color `'#2b5c8f'`, label `'sin(x)'`
2. $y_2 = \cos(x)$: Dashed line (`'--'`), color `'#e27d60'`, label `'cos(x)'`
3. $y_3 = \sin(x) \cdot e^{-0.2x}$: Dotted line (`':'`), color `'#2ca02c'`, linewidth 2.2, label `'Damped'`
Add an axis title, legend with frame at upper right, and grid.


```python
x = np.linspace(0, 10, 150)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.exp(-0.2 * x)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
ax.plot(x, y1, linestyle='-', color='#2b5c8f', linewidth=2, label=r'$\sin(x)$')
ax.plot(x, y2, linestyle='--', color='#e27d60', linewidth=2, label=r'$\cos(x)$')
ax.plot(x, y3, linestyle=':', color='#2ca02c', linewidth=2.2, label='Damped')

ax.set_title('Trigonometric Waveforms & Damped Oscillator', fontweight='bold')
ax.set_xlabel('Time (t)')
ax.set_ylabel('Amplitude')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='upper right', frameon=True)
plt.show()
```

### Problem 4: Line Plot with Custom Markers

### Problem 5: Shaded Uncertainty Band (`fill_between`)

Simulate an experimental process:
- $t \in [0, 10]$ with 100 points
- Mean trajectory: $\mu(t) = 10 + 2t - 0.15 t^2$
- Uncertainty: $\sigma(t) = 0.5 + 0.1t$
Plot the mean curve in navy, and shade the $\mu \pm 1.5\sigma$ band using `ax.fill_between()` with `alpha=0.25`.


```python
t = np.linspace(0, 10, 100)
mu = 10 + 2 * t - 0.15 * t**2
sigma = 0.5 + 0.1 * t

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
ax.plot(t, mu, color='#1f77b4', linewidth=2.2, label=r'Mean Trajectory $\mu(t)$')
ax.fill_between(t, mu - 1.5 * sigma, mu + 1.5 * sigma, color='#1f77b4', alpha=0.25,
                 label=r'Uncertainty Band ($\mu \pm 1.5\sigma$)')
ax.set_title('Process Trajectory with Shaded Uncertainty Band', fontweight='bold')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Process Metric')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='upper left')
plt.show()
```

### Problem 6: 2D Scatter Plot with Proportional Marker Sizes (Bubble Chart)

Generate 60 random points:
- `x`: uniform in $[10, 100]$
- `y`: $1.2x + \mathcal{N}(0, 15)$
- `sizes`: uniform in $[30, 300]$ points$^2$
Plot a scatter chart where marker area equals `sizes`, `color='#2b5c8f'`, `alpha=0.6`, with black edge color.


```python
rng = np.random.default_rng(42)
x = rng.uniform(10, 100, 60)
y = 1.2 * x + rng.normal(0, 15, 60)
sizes = rng.uniform(30, 300, 60)

fig, ax = plt.subplots(figsize=(8, 5), dpi=100)
ax.scatter(x, y, s=sizes, color='#2b5c8f', alpha=0.6, edgecolors='black', linewidths=0.7)
ax.set_title('Bubble Chart: Marker Size Encodes 3rd Variable', fontweight='bold')
ax.set_xlabel('Independent Variable X')
ax.set_ylabel('Dependent Variable Y')
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Problem 7: Continuous Colormap Scatter & Colorbar

Using the data from Problem 6, map a continuous fourth variable `z = x + 0.8 * y` to the marker color using `cmap='plasma'`.

### Problem 8: Categorical Scatter with Horizontal Jitter

Given three customer tiers with 50 satisfaction scores each:
- Bronze: $\mathcal{N}(65, 8)$
- Silver: $\mathcal{N}(78, 6)$
- Gold: $\mathcal{N}(89, 5)$
Plot the observations on categorical X-axis positions (0, 1, 2). Apply horizontal jitter $\mathcal{U}(-0.12, 0.12)$ to each point so points do not stack directly on top of each other.


```python
rng = np.random.default_rng(101)
scores = [
    rng.normal(65, 8, 50),
    rng.normal(78, 6, 50),
    rng.normal(89, 5, 50)
]
tiers = ['Bronze', 'Silver', 'Gold']
colors = ['#cd7f32', '#c0c0c0', '#ffd700']

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
for i, group in enumerate(scores):
    jitter = rng.uniform(-0.12, 0.12, len(group))
    ax.scatter(i + jitter, group, color=colors[i], edgecolors='black', alpha=0.7, s=35, label=tiers[i])

ax.set_xticks([0, 1, 2])
ax.set_xticklabels(tiers, fontweight='bold')
ax.set_title('Customer Satisfaction by Tier (Horizontal Jitter Applied)', fontweight='bold')
ax.set_ylabel('Satisfaction Score (0-100)')
ax.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

### Problem 10: Horizontal Bar Chart with Inverted Y-Axis

Given model names and F1 scores:
`models = ['DecisionTree', 'kNN', 'NaiveBayes', 'LogisticReg', 'RandomForest', 'XGBoost']`
`f1_scores = [0.72, 0.78, 0.75, 0.83, 0.91, 0.94]`
Sort the models by score ascending, create a horizontal bar chart (`ax.barh`), invert the Y-axis so the highest performer is at the top, and label bars with `ax.bar_label()`.


```python
models = np.array(['DecisionTree', 'kNN', 'NaiveBayes', 'LogisticReg', 'RandomForest', 'XGBoost'])
f1_scores = np.array([0.72, 0.78, 0.75, 0.83, 0.91, 0.94])

sorted_idx = np.argsort(f1_scores)
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
bars = ax.barh(models[sorted_idx], f1_scores[sorted_idx], color='#41b3a3', edgecolor='black', height=0.55)
ax.bar_label(bars, fmt='%.2f', padding=4, fontweight='bold', fontsize=9)
ax.invert_yaxis()
ax.set_xlim(0, 1.08)
ax.set_title('Classifier Performance (Sorted F1 Score)', fontweight='bold')
ax.set_xlabel('F1 Score')
ax.grid(axis='x', linestyle=':', alpha=0.5)
ax.spines[['top', 'right']].set_visible(False)
plt.show()
```

### Problem 11: Grouped Bar Chart with Coordinate Offsets

### Problem 13: Normalized Histogram with Automated Binning

Generate 1,500 samples from a normal distribution $\mathcal{N}(50, 10)$.
Plot a normalized histogram with `density=True` and `bins='auto'`.
Overlay the theoretical Gaussian PDF curve $f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$ in red.


```python
rng = np.random.default_rng(42)
samples = rng.normal(50, 10, 1500)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
count, bins, _ = ax.hist(samples, bins='auto', density=True, color='#2b5c8f', alpha=0.6, edgecolor='white', label='Empirical Histogram')

x_grid = np.linspace(15, 85, 200)
pdf = (1 / (10 * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x_grid - 50) / 10)**2)
ax.plot(x_grid, pdf, color='red', linewidth=2.2, label='Theoretical PDF')

ax.set_title('Normalized Histogram vs. Theoretical Normal PDF', fontweight='bold')
ax.set_xlabel('Value')
ax.set_ylabel('Probability Density')
ax.legend(frameon=True)
ax.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

### Problem 15: Box Plot with Custom Patch Colors & Notches

Generate four datasets with 100 samples each:
- D1: $\mathcal{N}(20, 3)$
- D2: $\mathcal{N}(25, 4)$
- D3: $\mathcal{N}(22, 6)$
- D4: $\mathcal{N}(30, 2)$
Create a boxplot with `patch_artist=True`, `notch=True`, custom box face colors, and labeled categories `['A', 'B', 'C', 'D']`.


```python
d1 = rng.normal(20, 3, 100)
d2 = rng.normal(25, 4, 100)
d3 = rng.normal(22, 6, 100)
d4 = rng.normal(30, 2, 100)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
bp = ax.boxplot([d1, d2, d3, d4], tick_labels=['A', 'B', 'C', 'D'], patch_artist=True, notch=True,
                medianprops=dict(color='white', lw=2))
palette = ['#4575b4', '#74add1', '#fdae61', '#f46d43']
for patch, color in zip(bp['boxes'], palette):
    patch.set_facecolor(color)
    patch.set_alpha(0.8)

ax.set_title('Notched Box Plots across 4 Experimental Treatments', fontweight='bold')
ax.set_ylabel('Observed Yield')
ax.grid(axis='y', linestyle=':', alpha=0.5)
plt.show()
```

### Problem 17: Multi-Panel 2x2 Grid with Flattening

Create a 2x2 grid of subplots using `plt.subplots(2, 2, figsize=(9, 6), layout='constrained')`.
Iterate over `axs.flatten()` to plot:
- Subplot 0: $x^2$
- Subplot 1: $x^3$
- Subplot 2: $\sqrt{x}$
- Subplot 3: $e^x$
Over $x \in [0, 4]$. Set unique titles and grid on each.


```python
x_vals = np.linspace(0, 4, 100)
funcs = [
    (x_vals**2, r'$y = x^2$', '#1f77b4'),
    (x_vals**3, r'$y = x^3$', '#ff7f0e'),
    (np.sqrt(x_vals), r'$y = \sqrt{x}$', '#2ca02c'),
    (np.exp(x_vals), r'$y = e^x$', '#d62728')
]

fig, axs = plt.subplots(2, 2, figsize=(9, 6), dpi=100, layout='constrained')
for ax, (y_data, title, col) in zip(axs.flatten(), funcs):
    ax.plot(x_vals, y_data, color=col, lw=2)
    ax.set_title(title, fontweight='bold')
    ax.grid(True, linestyle=':', alpha=0.5)
fig.suptitle('Mathematical Functions in a 2x2 Multi-Panel Grid', fontsize=13, fontweight='bold')
plt.show()
```

### Problem 18: Dual Y-Axes with `twinx()`

Plot a 30-day timeline:
- Left Y-axis (blue): Server CPU Load (%) ranging from 30% to 90%
- Right Y-axis (orange): Active User Connections ranging from 1,000 to 5,000
Combine legends from both axes into a single legend box in the upper-left.


```python
days_30 = np.arange(1, 31)
cpu_load = 50 + 25 * np.sin(days_30 / 4.0) + rng.normal(0, 3, 30)
connections = 2500 + 1500 * np.cos(days_30 / 5.0) + rng.normal(0, 100, 30)

fig, ax1 = plt.subplots(figsize=(9, 4.5), dpi=100)
l1 = ax1.plot(days_30, cpu_load, color='#1f77b4', lw=2, label='CPU Load (%)')
ax1.set_xlabel('Day')
ax1.set_ylabel('CPU Load (%)', color='#1f77b4', fontweight='bold')
ax1.tick_params(axis='y', labelcolor='#1f77b4')
ax1.grid(True, linestyle=':', alpha=0.5)

ax2 = ax1.twinx()
l2 = ax2.plot(days_30, connections, color='#ff7f0e', lw=2, linestyle='--', label='Active Connections')
ax2.set_ylabel('Active Connections', color='#ff7f0e', fontweight='bold')
ax2.tick_params(axis='y', labelcolor='#ff7f0e')

lines = l1 + l2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')
ax1.set_title('Server Telemetry: CPU Utilization vs User Connections', fontweight='bold')
plt.show()
```

### Problem 19: Logarithmic Scaling and Spine Removal

Given $x \in [1, 10]$ and $y = 10^x$:
- Plot $y$ vs $x$ with logarithmic Y-axis scale (`ax.set_yscale('log')`)
- Remove top and right spines using `ax.spines[['top', 'right']].set_visible(False)`
- Offset left and bottom spines outward by 6 points using `.set_position(('outward', 6))`


```python
x_log = np.linspace(1, 8, 50)
y_log = 10**x_log

fig, ax = plt.subplots(figsize=(7, 4.5), dpi=100)
ax.plot(x_log, y_log, color='#2b5c8f', lw=2.2, marker='o', markersize=4)
ax.set_yscale('log')
ax.set_title('Exponential Growth on Clean Logarithmic Scale', fontweight='bold')
ax.set_xlabel('Input Variable X')
ax.set_ylabel('Metric Y (log10 scale)')
ax.spines[['top', 'right']].set_visible(False)
ax.spines['left'].set_position(('outward', 6))
ax.spines['bottom'].set_position(('outward', 6))
ax.grid(True, which='both', linestyle=':', alpha=0.4)
plt.show()
```

### Problem 20: High-Resolution Publication Figure Export

Create a clean plot of $y = \cos(x) e^{-x/5}$ over $x \in [0, 15]$.
Save the figure to disk as `'publication_plot.png'` with `dpi=300`, `bbox_inches='tight'`, and `facecolor='white'`.
Verify the file exists, print its size in KB, and delete it to clean up.


```python
import os
x_pub = np.linspace(0, 15, 150)
y_pub = np.cos(x_pub) * np.exp(-x_pub / 5.0)

fig, ax = plt.subplots(figsize=(6, 3.5), dpi=100)
ax.plot(x_pub, y_pub, color='#2b5c8f', lw=2)
ax.set_title('Damped Cosine (300 DPI Export)', fontweight='bold')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.grid(True, linestyle=':', alpha=0.5)
ax.spines[['top', 'right']].set_visible(False)

filename = 'publication_plot.png'
fig.savefig(filename, dpi=300, bbox_inches='tight', facecolor='white')
size_kb = os.path.getsize(filename) / 1024
print(f"Export successful! File '{filename}' size: {size_kb:.2f} KB")
os.remove(filename)
plt.close(fig)
```
