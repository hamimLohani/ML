### Problem 3: Create a DataFrame from a Dictionary

Create a DataFrame `students_df` representing 5 students with the following columns:

### Problem 14: Detecting Missing Data

Consider this dataset with missing values:
```python
exam_data = {
    'Student': ['Sam', 'Alex', 'Jordan', 'Taylor', 'Morgan'],
    'Midterm': [85.0, np.nan, 90.0, 72.0, np.nan],
    'Final': [88.0, 79.0, np.nan, 68.0, 95.0],
    'Attendance': [95, 80, 85, np.nan, 90]
}
exam_df = pd.DataFrame(exam_data)
```
1. Count total missing values (`NaN`) per column.
2. Print the total number of missing values in the entire DataFrame.

````python
exam_data = {
    'Student': ['Sam', 'Alex', 'Jordan', 'Taylor', 'Morgan'],
    'Midterm': [85.0, np.nan, 90.0, 72.0, np.nan],
    'Final': [88.0, 79.0, np.nan, 68.0, 95.0],
    'Attendance': [95, 80, 85, np.nan, 90]
}
exam_df = pd.DataFrame(exam_data)

print("Missing values per column:")
print(exam_df.isna().sum())
print(f"\nTotal missing in whole table: {exam_df.isna().sum().sum()}")
````

### Problem 15: Imputing Missing Values

Fill the missing values in `exam_df`:
1. Fill missing `'Midterm'` scores with the average (mean) of the existing Midterm scores.
2. Fill missing `'Final'` scores with the median of existing Final scores.
3. Fill missing `'Attendance'` with `100`.

````python
exam_imputed = exam_df.copy()

exam_imputed['Midterm'] = exam_imputed['Midterm'].fillna(exam_imputed['Midterm'].mean())
exam_imputed['Final'] = exam_imputed['Final'].fillna(exam_imputed['Final'].median())
exam_imputed['Attendance'] = exam_imputed['Attendance'].fillna(100)

display(exam_imputed)
````

### Problem 17: Summary Statistics and Value Counts

Using `students_df`:

2. Count how many students belong to each `'Major'` using `.value_counts()`.

### Problem 18: GroupBy Single Column Aggregation

Group `students_df` by `'Major'` and compute the average `'GPA'` and average `'Age'` for each major.

### Problem 19: Named Aggregations with .groupby().agg()

Group `students_df` by `'Major'` and compute:
1. `'Student_Count'`: total number of students in that major (`count`)
2. `'Avg_GPA'`: average GPA in that major (`mean`)
3. `'Highest_GPA'`: highest GPA in that major (`max`)
Reset the index so `'Major'` becomes a regular column.

````python
agg_summary = students_df.groupby('Major').agg(
    Student_Count=('Name', 'count'),
    Avg_GPA=('GPA', 'mean'),
    Highest_GPA=('GPA', 'max')
).reset_index()

display(agg_summary)
````

### Problem 20: String Manipulation and Datetime Parsing

Given the following enrollment log:
```python
logs_data = {
    'Raw_Student': ['  ALICE  ', 'bob smith', ' CHARLIE brown ', 'diana  '],
    'Date_Enrolled': ['2026-01-10', '2026-02-15', '2026-03-01', '2026-03-22']
}
logs_df = pd.DataFrame(logs_data)
```
1. Clean `'Raw_Student'` by trimming whitespace and capitalizing each word to Title Case.
2. Convert `'Date_Enrolled'` to datetime objects.
3. Extract the `'Month_Name'` and `'Day_of_Week'` into new columns.

````python
logs_data = {
    'Raw_Student': ['  ALICE  ', 'bob smith', ' CHARLIE brown ', 'diana  '],
    'Date_Enrolled': ['2026-01-10', '2026-02-15', '2026-03-01', '2026-03-22']
}
logs_df = pd.DataFrame(logs_data)

# 1. Clean strings
logs_df['Clean_Student'] = logs_df['Raw_Student'].str.strip().str.title()

# 2. Convert to datetime
logs_df['Date_Enrolled'] = pd.to_datetime(logs_df['Date_Enrolled'])

# 3. Extract calendar features
logs_df['Month_Name'] = logs_df['Date_Enrolled'].dt.month_name()
logs_df['Day_of_Week'] = logs_df['Date_Enrolled'].dt.day_name()

display(logs_df[['Clean_Student', 'Date_Enrolled', 'Month_Name', 'Day_of_Week']])
````
