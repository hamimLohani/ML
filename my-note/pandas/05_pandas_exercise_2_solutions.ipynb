### Problem 1: Vectorized Conditional Columns with np.where

Given a DataFrame of employee performance scores:
```python
df = pd.DataFrame({
    'Employee': ['Amy', 'Brad', 'Chris', 'Dana', 'Eric'],
    'Score': [92, 68, 74, 85, 59]
})
```
Use `np.where` to create a new column `'Performance'`:
- `'Excellent'` if Score >= 85
- `'Satisfactory'` if Score >= 70 but < 85
- `'Needs Improvement'` if Score < 70

````python
df = pd.DataFrame({
    'Employee': ['Amy', 'Brad', 'Chris', 'Dana', 'Eric'],
    'Score': [92, 68, 74, 85, 59]
})

# Nested np.where
df['Performance'] = np.where(
    df['Score'] >= 85, 'Excellent',
    np.where(df['Score'] >= 70, 'Satisfactory', 'Needs Improvement')
)

display(df)
````

### Problem 3: Regex Pattern Extraction with .str.extract()

Given a DataFrame with messy contact strings:
```python
contacts = pd.DataFrame({
    'Raw_Contact': [
        'John Doe (555-1234)',
        'Jane Smith (555-5678)',
        'Bob Johnson (555-9012)'
    ]
})
```
Use `.str.extract()` with regex capture groups `r'^(.*?) \((.*?)\)'` to extract `'Name'` and `'Phone'` into separate columns.

````python
contacts = pd.DataFrame({
    'Raw_Contact': [
        'John Doe (555-1234)',
        'Jane Smith (555-5678)',
        'Bob Johnson (555-9012)'
    ]
})

pattern = r'^(?P<Name>.*?) \((?P<Phone>.*?)\)'
extracted = contacts['Raw_Contact'].str.extract(pattern)
display(extracted)
````

### Problem 4: Calculating Duration and Elapsed Days

Given subscription start and end dates:
```python
subs = pd.DataFrame({
    'User': ['UserA', 'UserB', 'UserC'],
    'Start': ['2025-01-10', '2025-03-01', '2025-06-15'],
    'End': ['2025-04-10', '2025-03-31', '2026-06-15']
})
```
1. Convert `'Start'` and `'End'` to datetime.
2. Compute `'Duration_Days'` as integer days elapsed.
3. Identify subscriptions lasting longer than 100 days.

````python
subs = pd.DataFrame({
    'User': ['UserA', 'UserB', 'UserC'],
    'Start': ['2025-01-10', '2025-03-01', '2025-06-15'],
    'End': ['2025-04-10', '2025-03-31', '2026-06-15']
})

subs['Start'] = pd.to_datetime(subs['Start'])
subs['End'] = pd.to_datetime(subs['End'])

subs['Duration_Days'] = (subs['End'] - subs['Start']).dt.days
subs['Over_100_Days'] = subs['Duration_Days'] > 100

display(subs)
````

### Problem 6: Relational Joins with pd.merge()

Given:
```python
employees = pd.DataFrame({
    'Emp_ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Dept_ID': [10, 20, 10, 30]
})

departments = pd.DataFrame({
    'Dept_ID': [10, 20, 40],
    'Dept_Name': ['Engineering', 'Marketing', 'Legal']
})
```
1. Perform an `inner` join on `'Dept_ID'`.
2. Perform a `left` join keeping all employees.

````python
employees = pd.DataFrame({
    'Emp_ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Dept_ID': [10, 20, 10, 30]
})

departments = pd.DataFrame({
    'Dept_ID': [10, 20, 40],
    'Dept_Name': ['Engineering', 'Marketing', 'Legal']
})

inner_res = pd.merge(employees, departments, on='Dept_ID', how='inner')
print("Inner Join (Only matching departments):")
display(inner_res)

left_res = pd.merge(employees, departments, on='Dept_ID', how='left')
print("\nLeft Join (All employees preserved):")
display(left_res)
````

### Problem 7: Finding Unmatched Records (Anti-Join Pattern)

Using `employees` and `departments` from Problem 6, identify employees who belong to a department that does NOT exist in the `departments` table (e.g. Diana in Dept 30).
*Hint:* Use `how='left'` with `indicator=True`.

````python
merged_indicator = pd.merge(employees, departments, on='Dept_ID', how='left', indicator=True)
print("Merged with indicator:")
display(merged_indicator)

unmatched_employees = merged_indicator[merged_indicator['_merge'] == 'left_only']
print("\nEmployees without matching Department record:")
display(unmatched_employees[['Emp_ID', 'Name', 'Dept_ID']])
````

### Problem 9: Unpivoting Data with pd.melt()

Given wide data where months are columns:
```python
wide_df = pd.DataFrame({
    'Product': ['Laptop', 'Mouse'],
    'Jan_Sales': [20, 150],
    'Feb_Sales': [25, 180],
    'Mar_Sales': [30, 210]
})
```
Use `pd.melt()` to reshape this into a tidy table with columns: `'Product'`, `'Month'`, and `'Sales_Units'`.

````python
wide_df = pd.DataFrame({
    'Product': ['Laptop', 'Mouse'],
    'Jan_Sales': [20, 150],
    'Feb_Sales': [25, 180],
    'Mar_Sales': [30, 210]
})

long_df = pd.melt(
    wide_df,
    id_vars=['Product'],
    value_vars=['Jan_Sales', 'Feb_Sales', 'Mar_Sales'],
    var_name='Month',
    value_name='Sales_Units'
)

display(long_df)
````

### Problem 10: Ranking Values with .rank()

Given a leaderboard:
```python
runners = pd.DataFrame({
    'Runner': ['Alex', 'Brian', 'Carlos', 'David', 'Evan'],
    'Time_Sec': [54.2, 52.8, 56.1, 52.8, 58.0]
})
```
Add a `'Rank'` column ranking runners from fastest (shortest time = Rank 1) to slowest, using `method='min'`.

````python
runners = pd.DataFrame({
    'Runner': ['Alex', 'Brian', 'Carlos', 'David', 'Evan'],
    'Time_Sec': [54.2, 52.8, 56.1, 52.8, 58.0]
})

runners['Rank'] = runners['Time_Sec'].rank(ascending=True, method='min').astype(int)
runners = runners.sort_values(by='Rank')
display(runners)
````

### Problem 11: Cumulative Running Metrics (.cumsum & .cummax)

Given daily revenue for a store:
```python
financials = pd.DataFrame({
    'Day': [1, 2, 3, 4, 5],
    'Daily_Revenue': [1000, 1500, 800, 2200, 1800]
})
```
1. Add `'Running_Total'` showing cumulative sum of revenue.
2. Add `'Peak_Daily_Revenue'` showing the highest single-day revenue observed up to that day.

````python
financials = pd.DataFrame({
    'Day': [1, 2, 3, 4, 5],
    'Daily_Revenue': [1000, 1500, 800, 2200, 1800]
})

financials['Running_Total'] = financials['Daily_Revenue'].cumsum()
financials['Peak_Daily_Revenue'] = financials['Daily_Revenue'].cummax()

display(financials)
````

### Problem 13: Memory Optimization with Categorical Dtypes

Create a DataFrame of 100,000 rows with a repetitive categorical column `'City'` randomly chosen from `['New York', 'London', 'Paris', 'Tokyo']`.
1. Check memory usage with `.memory_usage(deep=True)`.
2. Convert `'City'` to `category` dtype.
3. Measure memory reduction.

````python
n_rows = 100_000
cities = np.random.choice(['New York', 'London', 'Paris', 'Tokyo'], size=n_rows)
df_large = pd.DataFrame({'City': cities})

mem_before = df_large['City'].memory_usage(deep=True)
df_large['City'] = df_large['City'].astype('category')
mem_after = df_large['City'].memory_usage(deep=True)

print(f"Memory before (object):   {mem_before:>10,} bytes")
print(f"Memory after (category): {mem_after:>10,} bytes")
print(f"Reduction:               {mem_before / mem_after:.1f}x smaller!")
````

### Problem 14: Discretization & Binning with pd.cut()

Given customer ages:
```python
ages_df = pd.DataFrame({

```
Use `pd.cut()` to segment `'Age'` into four bins: `[0, 18, 35, 60, 100]` with labels `['Youth', 'Young Adult', 'Middle Aged', 'Senior']`.

### Problem 15: Fluent End-to-End Method Chaining Pipeline

Given raw store orders:
```python
raw_orders = pd.DataFrame({
    'Order_ID': [1, 2, 3, 4, 5],
    'Status': ['Completed', 'Cancelled', 'Completed', 'Completed', 'Refunded'],
    'Amount': [150.0, 20.0, 320.0, 85.0, 45.0],
    'Tax_Exempt': [False, False, True, False, True]
})
```
Build a single fluent chained expression (enclosed in parentheses) that:
1. Filters for only `'Completed'` orders
2. Adds a new column `'Tax'` ($10\%$ of Amount if not Tax_Exempt, else $0$)
3. Adds `'Total_Charge'` (Amount + Tax)
4. Sorts by `'Total_Charge'` descending
5. Resets the index

````python
raw_orders = pd.DataFrame({
    'Order_ID': [1, 2, 3, 4, 5],
    'Status': ['Completed', 'Cancelled', 'Completed', 'Completed', 'Refunded'],
    'Amount': [150.0, 20.0, 320.0, 85.0, 45.0],
    'Tax_Exempt': [False, False, True, False, True]
})

clean_pipeline = (
    raw_orders
    .query("Status == 'Completed'")
    .assign(
        Tax=lambda d: np.where(d['Tax_Exempt'], 0.0, d['Amount'] * 0.10),
        Total_Charge=lambda d: d['Amount'] + d['Tax']
    )
    .sort_values(by='Total_Charge', ascending=False)
    .reset_index(drop=True)
)

display(clean_pipeline)
````
