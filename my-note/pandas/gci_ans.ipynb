#### <Practice Question 3-5>

Let's examine the relationship between `Score` and `Members` in the `anime_data_extracted` dataset.

(1) Group `Scores` into fixed rating ranges and save the result as `score_tiers`:
1. 0.0~6.0: 'Low'
2. 6.0~7.5: 'Average'
3. 7.5~8.5: 'High'
4. 8.5~10.0: 'Very High'

```python
# Convert the Score column to type float first
anime_data_extracted["Score"] = anime_data_extracted["Score"].astype(float)

scores = [0.0,6.0,7.5,8.5,10.0]
score_tier_names = ['Low','Average','High','Very High']
score_tiers = pd.cut(anime_data_extracted["Score"], scores, labels=score_tier_names)
```

(2) Divide the member counts into 4 equal-sized popularity groups and save the result as `popularity_tiers`. Use the names 'Niche', 'Moderate', 'Popular', and 'Viral Hit' from lowest to highers.

```python
popularity_tier_names = ['Niche','Moderate','Popular','Viral Hit']
popularity_tiers = pd.qcut(anime_data_extracted['Members'],4,labels=popularity_tier_names)
```

Run this code to display a 4 by 4 frequency grid showing how many anime fall into each score and popularity combination.

```python
# Turns row categories into a full 2D table
pd.DataFrame({"Score": score_tiers, "Members": popularity_tiers}).value_counts().unstack()
```

#### <Practice Question 3-8>

We discussed many methods of handling missing data. Now let's observe how the choice of which method to use affects our analysis of the data. We are going to test these three methods on the `Score` column and see how it affects the mean value of `Score`.

1. Drop all rows with missing data
2. Fill all missing `Score` data with -1
3. Fill all missing `Score` data with the mean value of the `Score` data we have

Run this cell to turn all of the values in `Score` into `float` so we can calculate the mean.


```python
anime_data = anime_data.replace('Unknown', np.nan)
cols_to_use2 = [
    "Name", "Score", "Type", "Premiered", "Studios", "Source",
]
anime_data_extracted_pracq = anime_data[cols_to_use2]
anime_data_extracted_pracq["Score"] = anime_data_extracted_pracq["Score"].astype(float)
```

(1) Calculate the mean `Score` after removing all rows with at least one missing value.

```python
anime_data_extracted_dropna = anime_data_extracted_pracq.dropna()
anime_data_extracted_dropna["Score"].mean()
```

(2) Calculate the mean `Score` after filling all missing values with -1.

```python
anime_data_extracted_filln1 = anime_data_extracted_pracq.fillna(-1)
anime_data_extracted_filln1["Score"].mean()
```

(3) Calculate the mean `Score` after filling all missing `Score` values with the mean `Score` value for the data we currently have.

```python
anime_data_extracted_fillmean = anime_data_extracted_pracq.fillna(anime_data_extracted_pracq["Score"].mean())
anime_data_extracted_fillmean["Score"].mean()
```

As we can see, the method we choose to handle the missing values greatly affects the mean `Score` value. There is never an absolute correct method to choose, and it is your job as a data scientist to balance the trade offs from both the data science side and the domain side and to choose the method that makes the most sense.
